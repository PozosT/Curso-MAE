## Topics Covered This Week

### Topic 1 — Simple Linear Regression (SLR) — `15-1-Repaso_regresion_lineal.ipynb`

| Concept | Key formula / idea |
|---|---|
| SLR model | $Y_i = \beta_0 + \beta_1 x_i + \varepsilon_i$, $\varepsilon_i \sim \mathcal{N}(0, \sigma^2)$ |
| LINE assumptions | **L**inearity, **I**ndependence, **N**ormality, **E**qual variance |
| OLS estimators | $\hat{\beta}_1 = S_{xy}/S_{xx}$, $\hat{\beta}_0 = \bar{y} - \hat{\beta}_1\bar{x}$ |
| Coefficient of determination | $R^2 = SS_R / SS_T = 1 - SS_E/SS_T$ |

---

### Topic 2 — Multiple Linear Regression (MLR) — `15-1-Repaso_regresion_lineal.ipynb`

| Concept | Key formula / idea |
|---|---|
| MLR model | $\mathbf{y} = \mathbf{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon}$ |
| OLS estimators | $\hat{\boldsymbol{\beta}} = (\mathbf{X}^\top\mathbf{X})^{-1}\mathbf{X}^\top\mathbf{y}$ |
| Individual t-test | $t = \hat{\beta}_j / \text{SE}(\hat{\beta}_j)$ |
| Global F-test | $F = (SS_R/p) / (SS_E/(n-p-1))$ |
| Adjusted $R^2$ | $\bar{R}^2 = 1 - (1-R^2)\cdot(n-1)/(n-p-1)$ |
| Variance Inflation Factor | $\text{VIF}_j = 1/(1 - R_j^2)$; VIF > 10 indicates multicollinearity |

---

### Topic 3 — Logistic Regression — `15-2-Repaso_regresion_logistica.ipynb`

| Concept | Key formula / idea |
|---|---|
| Logit link | $\ln\!\left(\frac{\pi}{1-\pi}\right) = \mathbf{x}^\top\boldsymbol{\beta}$ |
| Probability | $\hat{\pi} = 1/(1+e^{-\mathbf{x}^\top\hat{\boldsymbol{\beta}}})$ |
| Odds ratio | $\text{OR}_j = e^{\hat{\beta}_j}$ for one-unit increase in $x_j$ |
| Estimation | Maximum likelihood (no closed form) |
| Model fit | Deviance, Hosmer-Lemeshow test, AUC/ROC |

---

### Topic 4 — Poisson Regression and GLMs — `15-2-Repaso_regresion_logistica.ipynb`

| Concept | Key formula / idea |
|---|---|
| Poisson model | $\ln(\mu_i) = \mathbf{x}_i^\top\boldsymbol{\beta}$; $Y_i \sim \text{Poisson}(\mu_i)$ |
| Rate ratio | $\text{RR}_j = e^{\hat{\beta}_j}$ |
| Overdispersion | $\text{Var}(Y) \gg \mu$ → use Negative Binomial or quasi-Poisson |
| GLM framework | Exponential family + link function + linear predictor |

## Key Formulas Quick Reference

### Confidence and Prediction Intervals (SLR)

$$
\text{CI for } E[Y|x^*]: \quad \hat{y}^* \pm t_{\alpha/2,\,n-2} \cdot \hat{\sigma}\sqrt{\frac{1}{n} + \frac{(x^* - \bar{x})^2}{S_{xx}}}
$$

$$
\text{PI for } Y^*: \quad \hat{y}^* \pm t_{\alpha/2,\,n-2} \cdot \hat{\sigma}\sqrt{1 + \frac{1}{n} + \frac{(x^* - \bar{x})^2}{S_{xx}}}
$$

### Information Criteria

$$
\text{AIC} = -2\ell(\hat{\boldsymbol{\beta}}) + 2p \qquad \text{BIC} = -2\ell(\hat{\boldsymbol{\beta}}) + p\ln(n)
$$

Lower is better. BIC penalizes complexity more heavily for large $n$.

### Regularized Regression

| Method | Penalty | Effect |
|---|---|---|
| Ridge ($L_2$) | $\lambda\sum\beta_j^2$ | Shrinks coefficients, no exact zeros |
| Lasso ($L_1$) | $\lambda\sum|\beta_j|$ | Shrinks + sparse (exact zeros) |
| Elastic Net | $\alpha L_1 + (1-\alpha) L_2$ | Combines both |

Use cross-validation to select $\lambda$.
