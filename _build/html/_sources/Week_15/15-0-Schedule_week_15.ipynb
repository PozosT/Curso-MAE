## Temas de Esta Semana

### Tema 1 — Regresión Lineal Simple (RLS) — `15-1-Repaso_regresion_lineal.ipynb`

| Concepto | Fórmula / idea clave |
|---|---|
| Modelo RLS | $Y_i = \beta_0 + \beta_1 x_i + \varepsilon_i$, $\varepsilon_i \sim \mathcal{N}(0, \sigma^2)$ |
| Supuestos LINE | **L**inealidad, **I**ndependencia, **N**ormalidad, **E**quivarianza |
| Estimadores MCO | $\hat{\beta}_1 = S_{xy}/S_{xx}$, $\hat{\beta}_0 = \bar{y} - \hat{\beta}_1\bar{x}$ |
| Coeficiente de determinación | $R^2 = SS_R / SS_T = 1 - SS_E/SS_T$ |

---

### Tema 2 — Regresión Lineal Múltiple (RLM) — `15-1-Repaso_regresion_lineal.ipynb`

| Concepto | Fórmula / idea clave |
|---|---|
| Modelo RLM | $\mathbf{y} = \mathbf{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon}$ |
| Estimadores MCO | $\hat{\boldsymbol{\beta}} = (\mathbf{X}^\top\mathbf{X})^{-1}\mathbf{X}^\top\mathbf{y}$ |
| Prueba t individual | $t = \hat{\beta}_j / \text{SE}(\hat{\beta}_j)$ |
| Prueba F global | $F = (SS_R/p) / (SS_E/(n-p-1))$ |
| $R^2$ ajustado | $\bar{R}^2 = 1 - (1-R^2)\cdot(n-1)/(n-p-1)$ |
| Factor de Inflación de la Varianza | $\text{VIF}_j = 1/(1 - R_j^2)$; VIF > 10 indica multicolinealidad |

---

### Tema 3 — Regresión Logística — `15-2-Repaso_regresion_logistica.ipynb`

| Concepto | Fórmula / idea clave |
|---|---|
| Enlace logit | $\ln\!\left(\frac{\pi}{1-\pi}\right) = \mathbf{x}^\top\boldsymbol{\beta}$ |
| Probabilidad | $\hat{\pi} = 1/(1+e^{-\mathbf{x}^\top\hat{\boldsymbol{\beta}}})$ |
| Razón de momios | $\text{OR}_j = e^{\hat{\beta}_j}$ por un aumento de una unidad en $x_j$ |
| Estimación | Máxima verosimilitud (sin forma cerrada) |
| Ajuste del modelo | Devianza, prueba de Hosmer-Lemeshow, AUC/ROC |

---

### Tema 4 — Regresión de Poisson y GLMs — `15-2-Repaso_regresion_logistica.ipynb`

| Concepto | Fórmula / idea clave |
|---|---|
| Modelo Poisson | $\ln(\mu_i) = \mathbf{x}_i^\top\boldsymbol{\beta}$; $Y_i \sim \text{Poisson}(\mu_i)$ |
| Razón de tasas | $\text{RR}_j = e^{\hat{\beta}_j}$ |
| Sobredispersión | $\text{Var}(Y) \gg \mu$ → usar Binomial Negativa o quasi-Poisson |
| Marco GLM | Familia exponencial + función de enlace + predictor lineal |

## Referencia Rápida de Fórmulas Clave

### Intervalos de Confianza y de Predicción (RLS)

$$
\text{IC para } E[Y|x^*]: \quad \hat{y}^* \pm t_{\alpha/2,\,n-2} \cdot \hat{\sigma}\sqrt{\frac{1}{n} + \frac{(x^* - \bar{x})^2}{S_{xx}}}
$$

$$
\text{IP para } Y^*: \quad \hat{y}^* \pm t_{\alpha/2,\,n-2} \cdot \hat{\sigma}\sqrt{1 + \frac{1}{n} + \frac{(x^* - \bar{x})^2}{S_{xx}}}
$$

### Criterios de Información

$$
\text{AIC} = -2\ell(\hat{\boldsymbol{\beta}}) + 2p \qquad \text{BIC} = -2\ell(\hat{\boldsymbol{\beta}}) + p\ln(n)
$$

Un valor menor es mejor. El BIC penaliza la complejidad con más fuerza para $n$ grande.

### Regresión Regularizada

| Método | Penalización | Efecto |
|---|---|---|
| Ridge ($L_2$) | $\lambda\sum\beta_j^2$ | Contrae coeficientes, sin ceros exactos |
| Lasso ($L_1$) | $\lambda\sum|\beta_j|$ | Contrae + dispersa (ceros exactos) |
| Elastic Net | $\alpha L_1 + (1-\alpha) L_2$ | Combina ambos |

Use validación cruzada para seleccionar $\lambda$.
