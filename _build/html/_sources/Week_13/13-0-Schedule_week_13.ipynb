## Topic Overview

### 1. Poisson Distribution Review

The Poisson distribution models the number of events occurring in a fixed interval of time or space when events occur independently at a constant average rate $\lambda$. Its key property is that the mean equals the variance: $E[Y] = \text{Var}(Y) = \lambda$. We review the PMF, moment-generating function, and graphical shape for different values of $\lambda$.

### 2. Poisson Regression Model

Poisson regression is a Generalized Linear Model (GLM) in which:
- The **random component** is $Y_i \sim \text{Poisson}(\mu_i)$.
- The **systematic component** is the linear predictor $\eta_i = \mathbf{x}_i^\top \boldsymbol{\beta}$.
- The **link function** is the natural logarithm: $\log(\mu_i) = \eta_i$, ensuring $\mu_i > 0$.

Parameters are estimated by maximum likelihood via iteratively reweighted least squares (IRLS).

### 3. Interpretation of Coefficients — Rate Ratios

Because the link is the log, a unit increase in $x_j$ multiplies the expected count by $e^{\beta_j}$. This multiplicative factor is called a **rate ratio** (or incidence rate ratio, IRR). Values above 1 indicate an increase in the rate; values below 1 indicate a decrease.

### 4. Overdispersion and Negative Binomial Regression

The Poisson model assumes $E[Y] = \text{Var}(Y)$. Real count data often exhibit **overdispersion** (variance > mean), caused by unobserved heterogeneity or clustering. The **Negative Binomial** distribution relaxes this constraint by introducing a dispersion parameter $\alpha$: $\text{Var}(Y) = \mu + \alpha\mu^2$. When $\alpha \to 0$ the model reduces to Poisson.
