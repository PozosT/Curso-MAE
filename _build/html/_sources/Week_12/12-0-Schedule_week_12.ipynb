## Objective

This week focuses on evaluating logistic regression models for classification tasks. After fitting a logistic regression model, it is essential to measure how well the model performs and to compare competing models. We will study the most widely used evaluation metrics in binary classification: the confusion matrix, accuracy, precision, recall, F1-score, the ROC curve, and the AUC statistic. We will also introduce the concept of model calibration.

---

## Schedule

### Topic 1 — Confusion Matrix
- Definition of true positives (TP), true negatives (TN), false positives (FP), and false negatives (FN)
- How classification threshold determines the entries of the confusion matrix
- Visual representation and interpretation

### Topic 2 — Classification Metrics
- **Accuracy**: proportion of correctly classified observations
- **Precision**: of all predicted positives, how many are truly positive
- **Recall (Sensitivity)**: of all actual positives, how many are correctly identified
- **F1-Score**: harmonic mean of precision and recall
- **Specificity**: of all actual negatives, how many are correctly identified
- Trade-offs between precision and recall; choosing the right metric for the problem

### Topic 3 — ROC Curve and AUC
- Receiver Operating Characteristic (ROC) curve: sensitivity vs. 1 − specificity across all thresholds
- Area Under the Curve (AUC) as a threshold-independent measure of discriminative ability
- Interpreting AUC: random classifier (0.5) vs. perfect classifier (1.0)
- Comparing two or more models using ROC plots

### Topic 4 — Model Calibration
- Distinction between discrimination (ROC/AUC) and calibration
- Calibration curves (reliability diagrams): predicted probability vs. observed frequency
- Hosmer–Lemeshow goodness-of-fit test
- Practical implications of poor calibration
