## Objetivo

Ofrecer una visión unificada de la familia de **Modelos Lineales Generalizados (GLM)**, extendiendo el modelo lineal clásico a variables respuesta no normales. Introducir el **análisis de supervivencia** como marco para modelar datos de tiempo hasta el evento en presencia de censura, cubriendo el estimador de Kaplan-Meier y el modelo de regresión de Riesgos Proporcionales de Cox.

## Temas

### Parte 1 — Modelos Lineales Generalizados (`14-1-GLM_overview.ipynb`)

1. **El marco GLM**
   - Componente aleatorio: distribución de la respuesta de la familia exponencial
   - Componente sistemático: predictor lineal $\eta = \mathbf{X}\boldsymbol{\beta}$
   - Función de enlace: $g(\mu) = \eta$

2. **Familias GLM comunes**
   - Gaussiana / enlace identidad → regresión lineal clásica
   - Binomial / enlace logit → regresión logística
   - Poisson / enlace log → regresión para datos de conteo
   - Gamma / enlace inverso → datos continuos positivos con sesgo

3. **Selección y ajuste del modelo**
   - Devianza y devianza residual
   - Comparación de modelos basada en AIC
   - Elección de una familia GLM en la práctica

---

### Parte 2 — Análisis de Supervivencia (`14-2-Survival_analysis.ipynb`)

1. **Conceptos fundamentales**
   - Función de supervivencia $S(t) = P(T > t)$
   - Función de riesgo $h(t) = \lim_{\Delta t \to 0} \frac{P(t \le T < t+\Delta t \mid T \ge t)}{\Delta t}$
   - Tipos de censura (derecha, izquierda, intervalo)

2. **Estimador de Kaplan-Meier**
   - Estimador no paramétrico de límite del producto
   - Curvas KM y prueba log-rank para comparación entre grupos

3. **Modelo de Riesgos Proporcionales de Cox**
   - Modelo semi-paramétrico: $h(t|\mathbf{x}) = h_0(t)\exp(\mathbf{x}^\top\boldsymbol{\beta})$
   - Interpretación de las razones de riesgo
   - Estimación por verosimilitud parcial

4. **Verificación del supuesto de riesgos proporcionales**
   - Residuos de Schoenfeld
   - Gráfico log-log de supervivencia
