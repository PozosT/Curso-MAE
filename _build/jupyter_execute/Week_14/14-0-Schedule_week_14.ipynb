## Objective

Provide a unified overview of the **Generalized Linear Model (GLM)** family, extending the classical linear model to non-normal response variables. Introduce **survival analysis** as a framework for modeling time-to-event data in the presence of censoring, covering the Kaplan-Meier estimator and the Cox Proportional Hazards regression model.

## Topics

### Part 1 — Generalized Linear Models (`14-1-GLM_overview.ipynb`)

1. **The GLM framework**
   - Random component: response distribution from the exponential family
   - Systematic component: linear predictor $\eta = \mathbf{X}\boldsymbol{\beta}$
   - Link function: $g(\mu) = \eta$

2. **Common GLM families**
   - Gaussian / identity link → classical linear regression
   - Binomial / logit link → logistic regression
   - Poisson / log link → count data regression
   - Gamma / inverse link → positive continuous, skewed data

3. **Model selection and fit**
   - Deviance and residual deviance
   - AIC-based model comparison
   - Choosing a GLM family in practice

---

### Part 2 — Survival Analysis (`14-2-Survival_analysis.ipynb`)

1. **Core concepts**
   - Survival function $S(t) = P(T > t)$
   - Hazard function $h(t) = \lim_{\Delta t \to 0} \frac{P(t \le T < t+\Delta t \mid T \ge t)}{\Delta t}$
   - Types of censoring (right, left, interval)

2. **Kaplan-Meier estimator**
   - Non-parametric product-limit estimator
   - KM curves and log-rank test for group comparison

3. **Cox Proportional Hazards model**
   - Semi-parametric model: $h(t|\mathbf{x}) = h_0(t)\exp(\mathbf{x}^\top\boldsymbol{\beta})$
   - Interpretation of hazard ratios
   - Partial likelihood estimation

4. **Checking the proportional hazards assumption**
   - Schoenfeld residuals
   - Log-log survival plot
