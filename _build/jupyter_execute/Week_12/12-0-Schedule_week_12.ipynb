## Objetivo

Esta semana se enfoca en la evaluación de modelos de regresión logística para tareas de clasificación. Después de ajustar un modelo de regresión logística, es fundamental medir qué tan bien se desempeña el modelo y comparar modelos en competencia. Estudiaremos las métricas de evaluación más utilizadas en clasificación binaria: la matriz de confusión, la exactitud, la precisión, la exhaustividad, el F1-score, la curva ROC y el estadístico AUC. También introduciremos el concepto de calibración de modelos.

---

## Programa

### Tema 1 — Matriz de Confusión
- Definición de verdaderos positivos (TP), verdaderos negativos (TN), falsos positivos (FP) y falsos negativos (FN)
- Cómo el umbral de clasificación determina las entradas de la matriz de confusión
- Representación visual e interpretación

### Tema 2 — Métricas de Clasificación
- **Exactitud (Accuracy)**: proporción de observaciones clasificadas correctamente
- **Precisión (Precision)**: de todos los positivos predichos, cuántos son verdaderamente positivos
- **Exhaustividad (Recall/Sensibilidad)**: de todos los positivos reales, cuántos son identificados correctamente
- **F1-Score**: media armónica de la precisión y la exhaustividad
- **Especificidad**: de todos los negativos reales, cuántos son identificados correctamente
- Compromisos entre precisión y exhaustividad; elegir la métrica adecuada para el problema

### Tema 3 — Curva ROC y AUC
- Curva Característica Operativa del Receptor (ROC): sensibilidad vs. 1 − especificidad para todos los umbrales
- Área Bajo la Curva (AUC) como medida de capacidad discriminativa independiente del umbral
- Interpretación del AUC: clasificador aleatorio (0.5) vs. clasificador perfecto (1.0)
- Comparación de dos o más modelos mediante gráficos ROC

### Tema 4 — Calibración de Modelos
- Distinción entre discriminación (ROC/AUC) y calibración
- Curvas de calibración (diagramas de fiabilidad): probabilidad predicha vs. frecuencia observada
- Prueba de bondad de ajuste de Hosmer–Lemeshow
- Implicaciones prácticas de una calibración deficiente
