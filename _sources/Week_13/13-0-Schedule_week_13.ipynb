## Descripción de los Temas

### 1. Repaso de la Distribución de Poisson

La distribución de Poisson modela el número de eventos que ocurren en un intervalo fijo de tiempo o espacio cuando los eventos ocurren de forma independiente a una tasa media constante $\lambda$. Su propiedad clave es que la media es igual a la varianza: $E[Y] = \text{Var}(Y) = \lambda$. Se repasa la FMP, la función generadora de momentos y la forma gráfica para diferentes valores de $\lambda$.

### 2. Modelo de Regresión de Poisson

La regresión de Poisson es un Modelo Lineal Generalizado (MLG) en el que:
- La **componente aleatoria** es $Y_i \sim \text{Poisson}(\mu_i)$.
- La **componente sistemática** es el predictor lineal $\eta_i = \mathbf{x}_i^\top \boldsymbol{\beta}$.
- La **función de enlace** es el logaritmo natural: $\log(\mu_i) = \eta_i$, asegurando que $\mu_i > 0$.

Los parámetros se estiman por máxima verosimilitud mediante mínimos cuadrados reponderados iterativamente (IRLS).

### 3. Interpretación de Coeficientes — Razones de Tasas

Debido a que el enlace es el logaritmo, un incremento unitario en $x_j$ multiplica el conteo esperado por $e^{\beta_j}$. Este factor multiplicativo se denomina **razón de tasas** (o razón de tasas de incidencia, RTI). Valores superiores a 1 indican un aumento en la tasa; valores inferiores a 1 indican una disminución.

### 4. Sobredispersión y Regresión Binomial Negativa

El modelo de Poisson supone $E[Y] = \text{Var}(Y)$. Los datos de conteo reales frecuentemente exhiben **sobredispersión** (varianza > media), causada por heterogeneidad no observada o agrupamiento. La distribución **Binomial Negativa** relaja esta restricción introduciendo un parámetro de dispersión $\alpha$: $\text{Var}(Y) = \mu + \alpha\mu^2$. Cuando $\alpha \to 0$ el modelo se reduce a Poisson.
